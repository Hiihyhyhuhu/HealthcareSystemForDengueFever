```python
from ddgs import DDGS
import pandas as pd

with DDGS() as ddgs:
    results = list(
        ddgs.text(
            "dengue fever symptoms",
            max_results=8
        )
    )

print(results)

df = pd.DataFrame(results)
df[["title", "href"]]
```

```python
import trafilatura

for i, r in enumerate(results):
    print(f"\n===== WEBSITE {i+1} =====")

    downloaded = trafilatura.fetch_url(r["href"])

    if downloaded is None:
        print("FAILED")
    else:
        text = trafilatura.extract(downloaded)

        if text:
            print("SUCCESS")
            print(text[:200])
        else:
            print("NO TEXT")
```

```python
documents = []

for r in results:
    downloaded = trafilatura.fetch_url(r["href"])

    if downloaded is not None:
        text = trafilatura.extract(downloaded)

        if text:
            documents.append({
                "title": r["title"],
                "url": r["href"],
                "content": text
            })

print("Number of documents:", len(documents))
```

```python
for doc in documents:
    print(doc["title"])
    print(len(doc["content"]))
    print("-"*50)
```
